# AI Impact on Student Learning - EDA & Data Cleaning (Phase 1)

**Objective**: Predict student burnout risk levels (Low, Medium, High) based on academic performance, AI tool usage, traditional study habits, and anxiety/wellbeing indicators.

## Project Workflow
1. **Data Loading & Initial Inspection**
2. **Data Cleaning** (Handling missing values, duplicate removal, type conversion, IQR outlier detection & removal)
3. **Feature Engineering** (`GPA_Improvement`, `AI_Efficiency`, `Study_Balance`, `Dependency_Index`)

## Step 1: Imports and Setup

```python
import sys
import os
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loader import DataLoader
from src.data_cleaner import DataCleaner
from src.feature_engineering import FeatureEngineer
from src.eda import EDAVisualizer

%matplotlib inline
sns.set_theme(style="whitegrid")
```

## Step 2: Data Loading and Inspection

```python
raw_data_path = '../data/ai_student_learning_raw.csv'
loader = DataLoader(data_path=raw_data_path)
df_raw = loader.load_data()
inspection_report = loader.inspect_data(df_raw)
```

## Step 3: Data Cleaning

## Step 4: Feature Engineering
